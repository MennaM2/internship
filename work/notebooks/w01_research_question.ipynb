## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

### Problem Framing

* **The Decision:** Which client web pages are currently experiencing a genuine performance decline and require an editorial refresh versus pages that are performing normally?
* **The Action:** The content team allocates editorial resources to review, update, and optimize the flagged pages.
* **Cost of a Wrong Call:**
  * **False Positive (Flagging a healthy page):** Wastes expensive editor time and budget on updating content that didn't need changes.
  * **False Negative (Missing a declining page):** Allows valuable pages to continuously lose search position and organic traffic unnoticed, leading to lost client revenue.
